## 2.Uncertainty in Estimation

```python
# Define settings
MaxTrial = 2000
sampleSizeRange = np.linspace(100, 200, 40)
plotVar = np.zeros(len(sampleSizeRange))
```

```python
# Loop through different sample sizes
for sSize in range(len(sampleSizeRange)):
    numSamples = int(sampleSizeRange[sSize])
    vStrial = np.zeros(MaxTrial)

    # Run 2000 trials for each sample size
    for trial in range(MaxTrial):
        xx = np.random.randn(numSamples, 1) # Generate Gaussian numbers
        vStrial[trial] = np.var(xx)         # Calculate variance

    # Measure how much the variance values vary from each other
    plotVar[sSize] = np.var(vStrial)
```

```python
# Plotting the results
fig, ax = plt.subplots(figsize=(5, 4))
ax.plot(sampleSizeRange, plotVar, marker='o', color='purple', linewidth=2)
ax.set_title("Uncertainty in Estimation", fontsize=12)
ax.set_xlabel("Sample Size (Number of Data Points)", fontsize=11)
ax.set_ylabel("Variance of Estimated Variance", fontsize=11)
ax.grid(True)
plt.tight_layout()
plt.show()
```

## 3. Bi-variate Gaussian Distribution

```python
# a funciton to calculate the 2d gaussian density
def gauss2D(x, m, C):
    Ci = np.linalg.inv(C)
    dC = np.linalg.det(C)

    # Mathematical formula for bivariate normal distribution
    num = np.exp(-0.5 * np.dot((x - m).T, np.dot(Ci, (x - m))))
    den = 2 * np.pi * np.sqrt(dC)
    return num / den
```

```python
# Function to generate the meshgrid grid and Z density values
def twoDGaussianPlot(nx, ny, m, C):
    # Expanded boundaries to fit the new means properly
    x = np.linspace(-5, 8, nx)
    y = np.linspace(-5, 8, ny)
    X, Y = np.meshgrid(x, y, indexing='ij')
    Z = np.zeros([nx, ny])
    for i in range(nx):
        for j in range(ny):
            xvec = np.array([X[i, j], Y[i, j]])
            Z[i, j] = gauss2D(xvec, m, C)
    return X, Y, Z
```

```python
# definning the 3 distributions
nx, ny = 100, 100

# Distribution 1
m1 = np.array([2.4, 3.2])
C1 = np.array([[2, -1], [-1, 2]], np.float32)

# Distribution 2
m2 = np.array([1.2, 0.2])
C2 = np.array([[2, 0], [0, 4]], np.float32)

# Distribution 3
m3 = np.array([2.4, 3.2])
C3 = np.array([[2, 0], [0, 2]], np.float32)
```

```python
# Generate Data
X1, Y1, Z1 = twoDGaussianPlot(nx, ny, m1, C1)
X2, Y2, Z2 = twoDGaussianPlot(nx, ny, m2, C2)
X3, Y3, Z3 = twoDGaussianPlot(nx, ny, m3, C3)
```

## 4. Sampling from a multi-variate Gaussia

```python
# Define the target covariance matrix
C = np.array([[2.0, 1.0], [1.0, 2.0]])

# Factorize the matrix using Cholesky Decomposition
A = np.linalg.cholesky(C)
print("Verified factorization (A @ A.T):\n", A @ A.T)
```

```python
# Generate 5000 standard, independent Gaussian data points (X)
X = np.random.randn(5000, 2)

# Transform the data points using matrix multiplication (Y)
Y = X @ A.T  # Using A.T aligns properly with row-vector mapping

print("\nShapes of the datasets:")
print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

```python
# 5. Draw the scatter plots
fig, ax = plt.subplots(figsize=(6, 6))

# Cyan dots for the original circular data
ax.scatter(X[:, 0], X[:, 1], c="c", s=4, alpha=0.6, label="Original Data (X)")

# Magenta dots for the transformed slanted data
ax.scatter(Y[:, 0], Y[:, 1], c="m", s=4, alpha=0.6, label="Transformed Data (Y)")

# Set plot limits and styling
ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_title("Sampling from a Multi-variate Gaussian", fontsize=12)
ax.set_xlabel("X1 Axis", fontsize=10)
ax.set_ylabel("X2 Axis", fontsize=10)
ax.set_aspect('equal', adjustable='box')  # Ensures geometry isn't stretched
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

## 5.Distribution of Projections

```python
np.random.seed(42)
C = np.array([[2.0, 1.0], [1.0, 2.0]])
A = np.linalg.cholesky(C)
X = np.random.randn(5000, 2)
Y = X @ A.T  # Transformed Data
```

```python
# get the varience
nPoints = 100
pVars = np.zeros(nPoints)
thRange = np.linspace(0, 2 * np.pi, nPoints)

for n in range(nPoints):
    theta = thRange[n]
    # Direction vector u
    u = np.array([np.sin(theta), np.cos(theta)])
    # Project data onto vector u
    yp = Y @ u
    pVars[n] = np.var(yp)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(C)
print("Mathematical Results")
print("Eigenvalues of C:", eigenvalues)
print("Eigenvectors of C:\n", eigenvectors)
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(thRange, pVars, color='crimson', linewidth=2.5, label="Projected Variance")

# Maxima and the minima
ax.axhline(y=max(pVars), color='g', linestyle='--', label=f"Max Variance (~{max(pVars):.2f})")
ax.axhline(y=min(pVars), color='b', linestyle='--', label=f"Min Variance (~{min(pVars):.2f})")

ax.set_title("Variance of Projected Data vs. Theta ($\theta$)", fontsize=12)
ax.set_xlabel("Angle Theta (Radians)", fontsize=10)
ax.set_ylabel("Measured Variance", fontsize=10)
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_xticklabels(['0', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$'])
ax.grid(True, alpha=0.4)
ax.legend(loc="lower right")

plt.tight_layout()
plt.show()
```
